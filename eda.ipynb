```python
import os
import json
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings

warnings.filterwarnings("ignore")
%matplotlib inline
```

```python
random.seed(42)
```

```python
file_path = "VQA_RAD/train.json"
train_data = pd.read_json(file_path)
train_data.head()
```

```python
print(train_data['answer_type'].unique())
```

```python
closed_ended_train_data = train_data[train_data['answer_type'] == 'CLOSED']
open_ended_train_data = train_data[train_data['answer_type'] == 'OPEN'] 

print(closed_ended_train_data['answer_type'].count())
print(open_ended_train_data['answer_type'].count())
```

```python
print(closed_ended_train_data['answer'].unique())
```

```python
print(open_ended_train_data['answer'][0])
```

```python
closed_ended_train_data['answer'] = closed_ended_train_data['answer'].str.lower()
print(closed_ended_train_data['answer'].unique())
```

```python
print(closed_ended_train_data['answer_type'].count())
```

```python
closed_ended_train_data['answer'] = closed_ended_train_data['answer'].str.strip()
print(closed_ended_train_data['answer'].unique())
```

```python
print(closed_ended_train_data.columns.to_list())
```

```python
len(train_data.columns)
```

```python
len(closed_ended_train_data.columns)
```

```python
no_ans_data = closed_ended_train_data[closed_ended_train_data['answer'] == 'no']
print(no_ans_data.count())
len(no_ans_data.columns)
```

```python
yes_ans_data = closed_ended_train_data[closed_ended_train_data['answer'] == 'yes']
print(yes_ans_data.count())
len(yes_ans_data.columns)
```

```python
binary_data = no_ans_data.merge(yes_ans_data, how='outer')
binary_data.head()
```

```python
print(binary_data.count())
print(binary_data['answer'].unique())
len(binary_data.columns)
```

```python
# save binary data as JSON file
with open(os.path.join(VQA_RAD, "binary_train.json"), "w") as f:
    json.dump(binary_data, f, indent=2)
```
